### 1.1 Overview of CDC-NHANES Dataset

The CDC-NHANES (National Health and Nutrition Examination Survey) dataset is a complete and solid source of health-related information derived from a nationally representative sample of the US population. The NHANES is administered by the Centers for Disease Control and Prevention (CDC), and it incorporates a wide range of demographic, clinical, laboratory, dietary, and lifestyle variables to provide valuable information about population health trends, disease prevalence, and health outcomes.

For this analysis, five out of six different files from the CDC-NHANES dataset will be used:

- **Demographic Data (`demographic.csv`)**: Captures essential demographic information, including age, gender, ethnicity, socioeconomic status, and other important demographic indicators.
- **Laboratory Data (`laboratory.csv`)**: Contains laboratory test results related to various health conditions, such as cholesterol levels and blood glucose measurements.
- **Examination Data (`examination.csv`)**: Provides extensive physical examination data, including body measurements and blood pressure readings.
- **Dietary Data (`diet.csv`)**: Records detailed dietary intake and nutritional habits over specified periods.
- **Questionnaire Data (`questionnaire.csv`)**: Includes self-reported information on medical history, health conditions, lifestyle habits, and behavioral factors.

These datasets are linked using the unique respondent identifier (SEQN), allowing for a combined evaluation of individual health profiles.

#### 2.2.1 Creating the Target Variable 'Diabetes_Status'

Upon closer examination, we notice that missing values in `Prediabetes_or_Higher` often coincide with available information in `Doctor_Diagnosed_Diabetes`. To leverage all available data, we decide to combine these variables into a ccomplete target variable, `Diabetes_Status`.

**Variable Definitions:**

- **Doctor_Diagnosed_Diabetes (DIQ010):**
  - 1: Yes
  - 2: No
  - 3: Borderline
  - 7: Refused
  - 9: Don't know

- **Prediabetes_or_Higher (DIQ160):**
  - 1: Yes
  - 2: No
  - 7: Refused
  - 9: Don't know

**Mapping Logic:**

- If `Doctor_Diagnosed_Diabetes` is 1 (Yes), then `Diabetes_Status` is `Diabetes`.
- If `Doctor_Diagnosed_Diabetes` is 3 (Borderline) or `Prediabetes_or_Higher` is 1 (Yes), then `Diabetes_Status` is `Prediabetes`.
- If `Doctor_Diagnosed_Diabetes` is 2 (No) and `Prediabetes_or_Higher` is 2 (No) or missing, then `Diabetes_Status` is `Normal`.
- If both variables are missing or have codes 7 or 9 (Refused or Don't know), the status is set as missing.


```python
# Function to create 'Diabetes_Status' based on questionnaire responses
def define_diabetes_status(row):
    doctor_diagnosed_diabetes = row['Doctor_Diagnosed_Diabetes']
    prediabetes_status = row['Prediabetes_or_Higher']
    
    # Handle missing or unknown values (7: Refused, 9: Don't know)
    if doctor_diagnosed_diabetes in [7, 9]:
        doctor_diagnosed_diabetes = np.nan
    if prediabetes_status in [7, 9]:
        prediabetes_status = np.nan
    
    # If diagnosed with diabetes
    if doctor_diagnosed_diabetes == 1:
        return 'Diabetes'
    # If diagnosed as borderline (considered prediabetes)
    elif doctor_diagnosed_diabetes == 3:
        return 'Prediabetes'
    # If not diagnosed with diabetes
    elif doctor_diagnosed_diabetes == 2:
        # Check prediabetes status
        if prediabetes_status == 1:
            return 'Prediabetes'
        elif prediabetes_status == 2 or pd.isnull(prediabetes_status):
            return 'Normal'
        else:
            return np.nan
    else:
        # If 'Doctor_Diagnosed_Diabetes' is missing
        if prediabetes_status == 1:
            return 'Prediabetes'
        elif prediabetes_status == 2:
            return 'Normal'
        else:
            # Both variables are missing or unknown
            return np.nan

# Apply the function to create 'Diabetes_Status'
data['Diabetes_Status'] = data.apply(define_diabetes_status, axis=1)

# Display the unique values of 'Diabetes_Status'
print("Unique values of 'Diabetes_Status':")    
print(data['Diabetes_Status'].value_counts(dropna=False))
```

We have 397 participants with missing 'Diabetes_Status'. For the purpose of our analysis, we decide to drop these rows.

However for the rest of the missing values, we will proceed using the MICE (Multiple Imputation by Chained Equations) method. MICE is advantageous as it models missing data iteratively based on interrelationships among all variables, ensuring consistency and retaining underlying patterns in the data.

MICE captures complex dependencies that simpler methods like mean/median or KNN cannot. Unlike KNN, it is computationally efficient for large datasets, and it avoids the bias introduced by mean/median imputation, ensuring more accurate and robust results. Also for the nutritional variables, with high missingess, it would be useful to use MICE because they are likely to be correlated with each other.

Before applying MICE imputation, scaling of numerical variables is necessary to ensure the imputation algorithm performs optimally. Since the variables requiring imputation are numerical, no additional encoding for categorical variables is required.

```python
# Original dataset
data_original = data.copy()

# Dataset after MICE imputation
data_mice = data_original.copy()

# Dataset after removal of missing values
data_remove = data_original.copy()
```

```python
def impute_mice(data, columns):
    from sklearn.experimental import enable_iterative_imputer
    from sklearn.impute import IterativeImputer

    scaler = StandardScaler()
    data_scaled = scaler.fit_transform(data[columns])

    imputer = IterativeImputer(max_iter=20, random_state=0)
    imputed_data = imputer.fit_transform(data_scaled)

    data_imputed = scaler.inverse_transform(imputed_data)
    data[columns] = data_imputed
    return data

def remove_missing(data, columns):
    data.dropna(subset=columns, inplace=True)
    return data
```

```python
# Columns to impute
cols_to_impute = ['Creatinine', 'Glycohemoglobin', 'Systolic_BP_Read1', 'Diastolic_BP_Read1', 'Total_Cholesterol', 'BMI', 'Waist_Circumference', 'Calories', 'Total_Carbohydrates', 'Total_Sugars', 'Fiber', 'Total_Fat', 'Saturated_Fatty_Acids', 'Income_Poverty_Ratio']

# Apply MICE imputation
data_mice = impute_mice(data_mice, cols_to_impute)

# Apply removal of missing values
data_remove = remove_missing(data_remove, cols_to_impute)
```

```python
data_imputed = data_mice.copy()
display_missing_data(data_imputed, title="Missing Data Overview (KNN Imputed Dataset)")
```

```python
def summarize_missing_removal(df, dataset_name):
    total_rows = len(df)
    remaining_rows = len(df.dropna())
    removed_rows = total_rows - remaining_rows
    percent_lost = (removed_rows / total_rows) * 100
    percent_retained = 100 - percent_lost
    
    print(f"Summary for {dataset_name}:")
    print(f"Total rows: {total_rows}")
    print(f"Rows retained after removing missing values: {remaining_rows} ({percent_retained:.2f}%)")
    print(f"Rows removed due to missing values: {removed_rows} ({percent_lost:.2f}%)")

# Summarize data loss for the dataset
summarize_missing_removal(data_original, "Original Dataset")
```

#### 3.2.2 Scaling and Normalization



Scaling techniques will be applide to the numerical variables to prepare them for clustering algorithms, which are sensitive to the scale of data. Algorithms like K-Means and DBSCAN use distance metrics (e.g., Euclidean distance). Features with larger scales can dominate the distance calculations. We use StandardScaler because it standardizes the features to have a mean of 0 and a standard deviation of 1, which is suitable for algorithms that assume standard normally distributed data.

Before applying the Standard Scaler, we will apply a skewness transformation to the data to make it more normally distributed. This can improve the performance of many machine learning algorithms that assume normality. By reducing skewness, the transformation helps in stabilizing variance and minimizing the impact of outliers.

The standard scaler then standardizes the features by removing the mean and scaling to unit variance. This ensures that each feature contributes equally to the model, preventing features with larger scales from dominating the learning process.

Combining these two steps can lead to better model performance and more reliable predictions.

```python
# List of numerical features
numeric_features = [
    'Age', 'Income_Poverty_Ratio', 'Systolic_BP_Read1', 'Diastolic_BP_Read1', 'BMI',
    'Creatinine', 'Glycohemoglobin', 'Total_Cholesterol', 'Calories', 'Fiber', 'Total_Sugars', 'Education_Level'
]

# Identify skewed features using skewness metric
from scipy.stats import skew

skewed_features = []
for feature in numeric_features:
    skewness = skew(data_encoded[feature].dropna())
    if skewness > 1 or skewness < -1:
        skewed_features.append(feature)
        print(f"{feature} is skewed (skewness = {skewness:.2f})")
```

```python
# Apply log transformation to skewed features
for feature in skewed_features:
    # Add a small constant to avoid log(0)
    data_encoded[feature] = np.log1p(data_encoded[feature]) #Computes natural log of 1 plus the input array, handling zeros and ensuring numerical stability.
    print(f"Applied log transformation to {feature}.")
```

```python
# Visualize distributions before and after log transformation
for feature in skewed_features:
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    
    # Original Distribution
    sns.histplot(data_imputed[feature], kde=True, ax=axes[0], color='skyblue')
    axes[0].set_title(f'Original Distribution of {feature}')
    
    # Distribution After Log Transformation
    transformed_feature = np.log1p(data_imputed[feature])
    sns.histplot(transformed_feature, kde=True, ax=axes[1], color='salmon')
    axes[1].set_title(f'{feature} After Log Transformation')
    
    plt.tight_layout()
    plt.show()
```

We can see that in most of them the skewnees has been reduced.


#### 3.3.1 K-Means Clustering



We can use methods like the Elbow Method and Silhouette Analysis to determine the optimal number of clusters.

```python
# Initialize the KMeans model
model = KMeans(random_state=42)

# Use the Elbow Method to find the optimal number of clusters
visualizer = KElbowVisualizer(model, k=(2, 10), metric='distortion', timings=False) #Tests the number of clusters from 2 to 15.
visualizer.fit(clustering_features)
visualizer.show()
```

Here k=(2, 10): It tests the number of clusters ranging from 2 to 10.

Metric='distortion': Uses the distortion score (sum of squared distances between data points and their cluster centroids) to evaluate cluster quality.


This suggests that  𝑘=5  is a suitable choice for the number of clusters, as adding more clusters beyond this point yields diminishing returns in terms of reducing the distortion score. Therefore, for KMeans clustering, k=5 appears to be the optimal number of clusters.

```python
# Define the range of cluster numbers to test
cluster_range = range(2, 6)

# Set up the figure with subplots
fig, axes = plt.subplots(len(cluster_range), 1, figsize=(8, len(cluster_range) * 4))

# Loop through each cluster size and create a Silhouette plot
for i, n_clusters in enumerate(cluster_range):
    # Initialize KMeans with the current number of clusters
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    
    # Create the Silhouette visualizer for the current cluster size
    visualizer = SilhouetteVisualizer(kmeans, colors='yellowbrick', ax=axes[i])
    visualizer.fit(clustering_features)
    visualizer.ax.set_title(f'Silhouette plot for k={n_clusters}')

# Adjust layout and show the plots
plt.tight_layout()
plt.show()
```

We chose \(k = 3\) because it balances high silhouette scores with meaningful clustering. While \(k = 2\) has a slightly higher score, it oversimplifies the data, and \(k > 3\) results in lower scores and poorly defined clusters. \(k = 3\) provides somewhat well-separated, interpretable clusters without overfitting.

```python
# Fit KMeans with the optimal number of clusters
optimal_k = 3
kmeans = KMeans(n_clusters=optimal_k, random_state=42)
cluster_labels = kmeans.fit_predict(clustering_features)

# Add cluster labels to the dataset
data_encoded['Cluster'] = cluster_labels

print(f"Clusters assigned. Number of clusters: {optimal_k}")
```

#### 3.3.2 Agglomerative Clustering


**Agglomerative Clustering** is a hierarchical clustering method that builds nested clusters by successively merging or splitting them based on distance criteria. Unlike K-Means clustering, which partitions data into a predefined number of clusters by minimizing within-cluster variance, Agglomerative Clustering creates a hierarchy of clusters without requiring you to specify the number of clusters upfront.


We will use the same standardized features used in K-Means clustering, excluding the target variable and any identifiers.

```python
# Exclude target variable and identifiers
clustering_features = data_encoded.drop(columns=['Participant_ID', 'Diabetes_Status', 'Diabetes_Status_Encoded', 'Cluster'])

print(f"Number of features used for clustering: {clustering_features.shape[1]}")
```

We compute the linkage matrix using Ward's method and plot the dendrogram to visualize the hierarchical clustering. This helps in determining the optimal number of clusters.

```python
# Compute linkage matrix using Ward's method
linked = linkage(clustering_features, method='ward')

# Plot dendrogram
plt.figure(figsize=(12, 8))
dendrogram(linked, truncate_mode='lastp', p=30, leaf_rotation=90, leaf_font_size=12)
plt.title('Hierarchical Clustering Dendrogram', fontsize=16)
plt.xlabel('Cluster Size', fontsize=12)
plt.ylabel('Distance', fontsize=12)
plt.show()
```

- **`method='ward'`**: 
  - Uses Ward's linkage criterion, which minimizes the variance within clusters.
  - This method is effective for quantitative data and tends to produce clusters of similar size.


Thus, based on the dendrogram:
- **Optimal Number of Clusters**: 3

```python
# Fit Agglomerative Clustering
n_clusters = 3  # Chosen based on the dendrogram
agglo = AgglomerativeClustering(n_clusters=n_clusters, linkage='ward')

# Predict cluster labels
agglo_labels = agglo.fit_predict(clustering_features)

# Add cluster labels to the dataset
data_encoded['Agglomerative_Cluster'] = agglo_labels

print(f"Clusters assigned. Number of clusters: {n_clusters}")
```

### 3.4 Association Rule Mining

Association Rule Mining uncovers interesting relationships between variables in large datasets. In the context of this health dataset, it can reveal patterns that contribute to the risk of diabetes. This section applies the Apriori algorithm to discover such associations, focusing on variables that are clinically significant.

#### 3.4.1 Preparing Data for Association Rules

We discretize key health-related variables into meaningful categories.

```python
#let's look at the summary of data_imputed

data_imputed.describe()
```

```python
# Use the data_imputed dataset
association_data = data_imputed.copy()

# Discretize Age into age groups
association_data['Age_Group'] = pd.cut(
    association_data['Age'], 
    bins=[19, 39, 59, 80], 
    labels=['Young Adult', 'Middle Age', 'Senior']
)

# Discretize BMI into categories
association_data['BMI_Category'] = pd.cut(
    association_data['BMI'], 
    bins=[0, 18.5, 25, 30, np.inf], 
    labels=['Underweight', 'Normal', 'Overweight', 'Obese']
)

# Discretize Systolic Blood Pressure into categories
association_data['Systolic_BP_Category'] = pd.cut(
    association_data['Systolic_BP_Read1'], 
    bins=[0, 120, 130, 140, np.inf], 
    labels=['Normal', 'Elevated', 'Stage 1 Hypertension', 'Stage 2 Hypertension']
)

# Discretize Glycohemoglobin (HbA1c) into categories
association_data['HbA1c_Category'] = pd.cut(
    association_data['Glycohemoglobin'], 
    bins=[0, 5.7, 6.5, np.inf], 
    labels=['Normal', 'Prediabetes', 'Diabetes']
)

# Discretize Total Cholesterol into categories
association_data['Cholesterol_Level'] = pd.cut(
    association_data['Total_Cholesterol'], 
    bins=[0, 200, 240, np.inf], 
    labels=['Desirable', 'Borderline High', 'High']
)

# Select relevant columns for mining
association_columns = ['Diabetes_Status', 'Gender', 'Family_History_Diabetes', 
                                 'Age_Group', 'BMI_Category', 'Systolic_BP_Category', 
                                 'HbA1c_Category', 'Cholesterol_Level', 'Race_Ethnicity']

# Subset the data
association_data = association_data[association_columns]
```


#### 3.4.2 Applying Association Rule Mining
We convert the selected data into a one-hot encoded format suitable for the Apriori algorithm and generate frequent itemsets and association rules.

```python
# Convert categorical features into one-hot encoded format
transactional_data = pd.get_dummies(association_data)

# Generate frequent itemsets with minimum support
frequent_itemsets = apriori(transactional_data, min_support=0.05, use_colnames=True)

# Generate association rules with minimum confidence
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.6)

# Filter rules based on lift
rules = rules[rules['lift'] > 1.2]

# Display top rules
print("Top association rules based on confidence and lift:")
rules_sorted = rules.sort_values(by=['lift', 'confidence'], ascending=False)
display(rules_sorted.head(5))
```

We select the top three significant association rules based on support, confidence, and lift, and interpret each rule in the context of diabetes risk.

#### 4.1.1 Evaluation Metrics and Visualization Tools

To assess the performance of our models, we'll use the following metrics:

- **Accuracy**: The proportion of correct predictions.
- **Precision**: The ratio of true positives to the sum of true and false positives.
- **Recall**: The ratio of true positives to the sum of true positives and false negatives.
- **F1-Score**: The harmonic mean of precision and recall.
- **Confusion Matrix**: Provides insight into the types of errors being made.
- **ROC Curve and AUC**: Measures the ability of the model to distinguish between classes.

We define functions to visualize these metrics:

```python
def plot_metrics(y_test, y_pred, labels=['Normal', 'Prediabetes', 'Diabetes']):
    '''Plot Confusion Matrix and Classification Report'''
    cm = metrics.confusion_matrix(y_test, y_pred)
    cr = pd.DataFrame(metrics.classification_report(y_test, y_pred, digits=3, output_dict=True)).T
    cr.drop(columns='support', inplace=True)
    
    fig, ax = plt.subplots(1, 2, figsize=(15, 5))
    
    # Confusion Matrix
    sns.heatmap(cm, annot=True, fmt='d', cmap="Blues", ax=ax[0])
    ax[0].set_xlabel('Predicted labels', fontsize=14)
    ax[0].set_ylabel('True labels', fontsize=14)
    ax[0].set_title('Confusion Matrix', fontsize=16)
    ax[0].xaxis.set_ticklabels(labels)
    ax[0].yaxis.set_ticklabels(labels)
    
    # Classification Report
    sns.heatmap(cr.iloc[:-1, :].T, cmap='Blues', annot=True, fmt='.3f', ax=ax[1])
    ax[1].set_title('Classification Report', fontsize=16)
    ax[1].set_xticklabels(cr.index[:-1], rotation=45)
    plt.tight_layout()
    plt.show()


def plot_roc_auc(y_true, y_score, classes, n_classes, title='Receiver Operating Characteristic'):
    """
    Plots ROC curves for each class and computes AUC.

    Parameters:
    - y_true: Array-like of shape (n_samples,)
        True class labels.
    - y_score: Array-like of shape (n_samples, n_classes)
        Target scores, can either be probability estimates of the positive class
        or confidence values.
    - classes: List
        List of class names.
    - n_classes: int
        Number of classes.
    - title: str
        Title for the plot.
    """
    # Binarize the output
    y_true_binarized = label_binarize(y_true, classes=range(n_classes))
    
    # Compute ROC curve and ROC area for each class
    fpr = dict()
    tpr = dict()
    roc_auc = dict()
    for i in range(n_classes):
        fpr[i], tpr[i], _ = roc_curve(y_true_binarized[:, i], y_score[:, i])
        roc_auc[i] = auc(fpr[i], tpr[i])
    
    # Compute micro-average ROC curve and ROC area
    fpr["micro"], tpr["micro"], _ = roc_curve(y_true_binarized.ravel(), y_score.ravel())
    roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])
    
    # Plot all ROC curves
    plt.figure(figsize=(10, 8))
    colors = sns.color_palette("viridis", n_colors=n_classes)
    for i, color in zip(range(n_classes), colors):
        plt.plot(fpr[i], tpr[i], color=color, lw=2,
                 label='ROC curve of class {0} (AUC = {1:0.2f})'
                 ''.format(classes[i], roc_auc[i]))
    
    plt.plot([0, 1], [0, 1], 'k--', lw=2)  # Diagonal line
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate', fontsize=14)
    plt.ylabel('True Positive Rate', fontsize=14)
    plt.title(title, fontsize=16)
    plt.legend(loc="lower right", fontsize=12)
    plt.grid(True)
    plt.show()
```

### 4.2 Model 1: Logistic Regression

Logistic Regression is a widely used statistical model for binary and multiclass classification problems. It is favored for its simplicity, interpretability, and efficiency. Given our goal to predict diabetes risk levels and the importance of understanding feature contributions, Logistic Regression is a suitable starting point.

#### 4.2.1 Logistic Regression with Bayesian Optimization

The choice of Bayesian Optimization is given by:

- Efficiency: Bayesian Optimization is more sample-efficient than grid or random search.
- Global Optimization: It balances exploration and exploitation to find the global optimum.
- Practicality: Suitable for expensive functions where evaluation time is significant.

Hyperparameters to Tune:

- penalty: Regularization type ('l1', 'l2', 'elasticnet', 'none').
- C: Inverse of regularization strength (range: 0.01 to 10).
- solver: Algorithm to use ('liblinear' supports L1 regularization).

```python
# Define parameter space
param_space = {
    'penalty': Categorical(['l1', 'l2']),
    'C': Real(0.01, 10.0, prior='log-uniform'),
    'class_weight': Categorical([None, 'balanced']),
    'solver': Categorical(['liblinear'])
}

# Initialize Logistic Regression
lr = LogisticRegression(multi_class='auto', max_iter=1000, random_state=42)

# Bayesian Search with cross-validation
bayes_search = BayesSearchCV(
    estimator=lr,
    search_spaces=param_space,
    n_iter=30,
    cv=5,
    n_jobs=-1,
    scoring='accuracy',
    random_state=42
)

# Fit the model
bayes_search.fit(X_train_val, y_train_val)

# Best parameters and score
print('Best hyperparameters:', bayes_search.best_params_)
print(f"Best cross-validation score: {bayes_search.best_score_:.4f}")
```

- C (Regularization Strength): A value of 0.0658 indicates strong regularization, which helps prevent overfitting by penalizing large coefficients more heavily.
- class_weight: None means no class weighting is applied, so the model treats all classes equally. This may impact performance if the classes are imbalanced.
- penalty: Using 'l1' applies Lasso regularization, which can lead to feature selection by shrinking some coefficients to zero.
- solver: The 'liblinear' solver supports 'l1' penalty and is efficient for smaller datasets.

We perform 30 iterations to explore the paramete space comprehensively and we do a 5-fold cross-validation to evaluate the model, balancing computational efficiency and reliable estiamtes.

```python
# Evaluate on test set
lr_best = bayes_search.best_estimator_
y_pred_lr = lr_best.predict(X_test)
accuracy = metrics.accuracy_score(y_test, y_pred_lr)
print(f"Test set Accuracy: {accuracy:.4f}")
plot_metrics(y_test, y_pred_lr)
plot_roc_auc(y_test, lr_best.predict_proba(X_test), classes=['Normal', 'Prediabetes', 'Diabetes'], n_classes=3)
```

The accuracy is not too bad on the test set, however it is evident the class unbalance in the data. The precision and recall are very low for the minority classes, completely 0 for prediabets.

#### 4.2.2 Logistic Regression with SMOTE and Bayesian Optimization

```python
# Apply SMOTE to the training data
smote = SMOTE(random_state=42)
X_resampled_lr, y_resampled_lr = smote.fit_resample(X_train_val, y_train_val)

# Bayesian Search with cross-validation on resampled data
bayes_search_smote = BayesSearchCV(
    estimator=lr,
    search_spaces=param_space,
    n_iter=30,
    cv=5,
    n_jobs=-1,
    scoring='accuracy',
    random_state=42
)

# Fit the model
bayes_search_smote.fit(X_resampled_lr, y_resampled_lr)

# Best parameters and score
print('Best hyperparameters (SMOTE):', bayes_search_smote.best_params_)
print(f"Best cross-validation score (SMOTE): {bayes_search_smote.best_score_:.4f}")
```

- C (Regularization Strength): A value of 0.2811 indicates moderate regularization, helping prevent overfitting by penalizing large coefficients.

- class_weight: Setting to 'balanced' adjusts weights inversely to class frequencies, addressing class imbalance and - improving minority class prediction.
- penalty: Using 'l2' applies Ridge regularization, promoting generalization by shrinking coefficients.
- solver: The 'liblinear' solver is efficient for small datasets and supports 'l2' penalty.

```python
# Evaluate on test set
lr_best_smote = bayes_search_smote.best_estimator_
y_pred_lr_smote = lr_best_smote.predict(X_test)
accuracy = metrics.accuracy_score(y_test, y_pred_lr_smote)
print(f"Test set Accuracy (SMOTE): {accuracy:.4f}")
plot_metrics(y_test, y_pred_lr_smote)
plot_roc_auc(y_test, lr_best_smote.predict_proba(X_test), classes=['Normal', 'Prediabetes', 'Diabetes'], n_classes=3)
```

The test set accuracy decreased to 72.7%, but the precision and recall for the 'Prediabetes' class improved significantly. ### 4.3 Model 2: Support Vector Machine (SVM)

Support Vector Machines are powerful classifiers that are effective in high-dimensional spaces and can model non-linear decision boundaries using kernel functions. SVMs are robust to overfitting, especially in cases where the number of dimensions exceeds the number of samples.

#### 4.3.1 SVM with Bayesian Optimization

Hyperparameters to Tune:
- C: Regularization parameter.
- kernel: Kernel type ('linear', 'rbf').
- gamma: Kernel coefficient.

```python
# Define parameter space
svm_param_space = {
    'C': Real(0.1, 10.0, prior='log-uniform'),
    'kernel': Categorical(['linear', 'rbf']),
    'gamma': Categorical(['scale', 'auto']),
    'class_weight': Categorical([None, 'balanced'])
}

# Initialize SVM classifier
svc = SVC(probability=True, random_state=42)

# Bayesian Search with cross-validation
svm_bayes_search = BayesSearchCV(
    estimator=svc,
    search_spaces=svm_param_space,
    n_iter=20,
    cv=5,
    n_jobs=-1,
    scoring='accuracy',
    random_state=42,
    verbose=True
)

# Fit the model
svm_bayes_search.fit(X_train_val, y_train_val)

# Best parameters and score
print('Best SVM hyperparameters:', svm_bayes_search.best_params_)
print(f"Best cross-validation score: {svm_bayes_search.best_score_:.4f}")
```

- C (Regularization Parameter): A value of 1.1866 indicates moderate regularization. A moderate C allows for some misclassifications but aims for a smoother decision boundary.

- class_weight: None means no class weighting is applied, so all classes are treated equally. In datasets with class imbalance, this might lead to the model favoring the majority class, potentially affecting minority class prediction performance.

- gamma: 'auto' sets gamma to 1 / n_features. Setting gamma='auto' adapts gamma based on the number of features, which can help in managing overfitting. 

- kernel: 'rbf' specifies the Radial Basis Function kernel, which is effective in capturing non-linear relationships in the data. The RBF kernel maps samples into a higher-dimensional space, allowing the model to find non-linear decision boundaries.

```python
# Evaluate on test set
svm_best = svm_bayes_search.best_estimator_
y_pred_svm = svm_best.predict(X_test)
accuracy = metrics.accuracy_score(y_test, y_pred_svm)
print(f"Test set Accuracy: {accuracy:.4f}")
plot_metrics(y_test, y_pred_svm)
plot_roc_auc(y_test, svm_best.predict_proba(X_test), classes=['Normal', 'Prediabetes', 'Diabetes'], n_classes=3)
```

For now it worked similarly to the logistic regression, with a low accuracy and low precision and recall for the minority classes. We will try to improve this with SMOTE. By looking at the ROC curve we can see that the model is not performing very well on the Prediabetes class, with an AUC a bit larger than 0.5, which means that the model is not able to distinguish between the classes.

#### 4.3.2 SVM with SMOTE and Bayesian Optimization

```python
# Resample training data
X_resampled_svm, y_resampled_svm = smote.fit_resample(X_train_val, y_train_val)

# Bayesian Search with cross-validation on resampled data
svm_bayes_search_smote = BayesSearchCV(
    estimator=svc,
    search_spaces=svm_param_space,
    n_iter=20,
    cv=5,
    n_jobs=-1,
    scoring='accuracy',
    random_state=42,
    verbose=True
)

# Fit the model
svm_bayes_search_smote.fit(X_resampled_svm, y_resampled_svm)

# Best parameters and score
print('Best SVM hyperparameters (SMOTE):', svm_bayes_search_smote.best_params_)
print(f"Best cross-validation score (SMOTE): {svm_bayes_search_smote.best_score_:.4f}")
```

- C: A higher value of 9.9444 implies lower regularization strength, allowing the model to fit the training data more closely. This can capture complex patterns but may risk overfitting if not managed carefully.

- class_weight: None indicates no class weighting is applied. Since SMOTE oversampling was used to balance the classes, additional class weighting may not be necessary.

- gamma: 'scale' sets gamma to 1 / (n_features * X.var()), adjusting automatically based on the data. This helps in controlling the influence of individual data points and can improve model performance on datasets with varying feature scales.

- kernel: 'rbf' specifies the Radial Basis Function kernel, as before.

```python
# Evaluate on test set
svm_best_smote = svm_bayes_search_smote.best_estimator_
y_pred_svm_smote = svm_best_smote.predict(X_test)
accuracy = metrics.accuracy_score(y_test, y_pred_svm_smote)
print(f"Test set Accuracy (SMOTE): {accuracy:.4f}")
plot_metrics(y_test, y_pred_svm_smote)
plot_roc_auc(y_test, svm_best_smote.predict_proba(X_test), classes=['Normal', 'Prediabetes', 'Diabetes'], n_classes=3)
```

We can see that with the SMOTE technique the model is now able to predict the minority classes better. The accuracy is lower than the previous model, but the precision and recall for the minority classes are higher. The F1-score for the 'Prediabetes' class is also higher, indicating a better balance between precision and recall. However, the model is still not performing very well, with a low accuracy and low precision and recall for the minority classes. We will have to explore other models to see if we can improve this.

**Discussion of Results**
- **Without SMOTE**:
  - The model ignores the minority "Prediabetes" class completely, leading to poor performance for this class but strong results for "Diabetes."

- **With SMOTE**:
  - SMOTE helps improve minority class ("Prediabetes") predictions slightly, with gains in F1-score and recall, though precision and overall AUC remain low.
  - Oversampling introduces noise, slightly reducing the AUC for "Prediabetes" and "Diabetes," though recall improves for the latter.

**Conclusion**
SMOTE addresses class imbalance, improving recall and F1 for minority classes, but at the cost of slight degradation in global ranking (AUC). 

### 4.4 Model 3: Random Forest

Random Forest is an ensemble learning method that operates by constructing multiple decision trees and aggregating their predictions. It improves predictive accuracy and controls overfitting by averaging multiple trees, reducing variance. Random Forests are known for their robustness and ability to handle high-dimensional data.

Hyperparameters to Tune:

- n_estimators: Number of trees.
- max_depth: Maximum depth of the trees.
- min_samples_split: Minimum number of samples required to split an internal node.
- max_features: Number of features to consider when looking for the best split.
- criterion: Function to measure the quality of a split.

```python
# Define parameter space
rf_param_space = {
    'n_estimators': Integer(100, 500),
    'max_depth': Integer(5, 50),
    'min_samples_split': Integer(2, 10),
    'max_features': Categorical(['sqrt', 'log2']),
    'criterion': Categorical(['gini', 'entropy']),
    'class_weight': Categorical([None, 'balanced'])
}

# Initialize Random Forest classifier
rf = RandomForestClassifier(random_state=42)

# Bayesian Search with cross-validation
rf_bayes_search = BayesSearchCV(
    estimator=rf,
    search_spaces=rf_param_space,
    n_iter=20,
    cv=5,
    n_jobs=-1,
    scoring='accuracy',
    random_state=42,
    verbose=True
)

# Fit the model
rf_bayes_search.fit(X_train_val, y_train_val)

# Best parameters and score
print('Best hyperparameters:', rf_bayes_search.best_params_)
print(f"Best cross-validation score: {rf_bayes_search.best_score_:.4f}")
```

- n_estimators: Using 100 trees provides a balance between model performance and computational efficiency. More trees can improve performance but increase computation time.

- max_depth: A max_depth of 28 allows trees to grow deep, enabling the model to capture complex patterns in the data. However, deeper trees can lead to overfitting if not controlled.

- min_samples_split: Setting min_samples_split to 7 means that a node must have at least 7 samples to be split. This introduces regularization by preventing splits that would result in nodes with very few samples, reducing overfitting.

- max_features: The 'log2' option selects features based on the logarithm (base 2) of the total number of features when considering splits. This ensures that each tree considers a subset of features, promoting diversity among trees and reducing correlation between them.

- criterion: Using the 'gini' impurity measure evaluates the quality of splits based on how well they separate the classes, favoring purer nodes. 

- class_weight: With None, no class weighting is applied, and all classes are treated equally. In datasets with class imbalance, this might affect the model's ability to predict minority classes effectively.

```python
# Evaluate on test set
rf_best = rf_bayes_search.best_estimator_
y_pred_rf = rf_best.predict(X_test)
accuracy = metrics.accuracy_score(y_test, y_pred_rf)
print(f"Test set Accuracy: {accuracy:.4f}")
plot_metrics(y_test, y_pred_rf)
plot_roc_auc(y_test, rf_best.predict_proba(X_test), classes=['Normal', 'Prediabetes', 'Diabetes'], n_classes=3)
```

- The model achieves high accuracy (86.84%) and performs well for the "Normal" and "Diabetes" classes, as reflected in their high F1-scores (0.928 and 0.711, respectively).
- However, the "Prediabetes" class is entirely ignored (precision, recall, and F1-score = 0), indicating that the model struggles to address the imbalance in the dataset.
- The ROC curve shows good discrimination for "Normal" (AUC = 0.87) and "Diabetes" (AUC = 0.93), but "Prediabetes" has poor AUC (0.71), demonstrating the model's inability to identify this class effectively.

Nothing new from what we have seen before.

#### 4.4.2 Random Forest with SMOTE and Bayesian Optimization

```python
# Resample training data
X_resampled_rf, y_resampled_rf = smote.fit_resample(X_train_val, y_train_val)

# Bayesian Search with cross-validation on resampled data
rf_bayes_search_smote = BayesSearchCV(
    estimator=rf,
    search_spaces=rf_param_space,
    n_iter=30,
    cv=5,
    n_jobs=-1,
    scoring='accuracy',
    random_state=42
)

# Fit the model
rf_bayes_search_smote.fit(X_resampled_rf, y_resampled_rf)

# Best parameters and score
print('Best hyperparameters (SMOTE):', rf_bayes_search_smote.best_params_)
print(f"Best cross-validation score (SMOTE): {rf_bayes_search_smote.best_score_:.4f}")
```

- n_estimators: Using 500 trees enhances the ensemble's stability and reduces variance, improving generalization to unseen data.
- max_depth: A max_depth of 50 allows trees to grow very deep, capturing intricate patterns in the data. With SMOTE balancing the classes, deeper trees are less likely to overfit on the majority class.

- min_samples_split: Setting this to 2 allows nodes to split even when they have two samples, maximizing the depth and complexity of each tree.

- max_features: The 'sqrt' option selects a subset of features equal to the square root of the total number of features at each split, promoting diversity among trees and reducing correlation within the forest.
- criterion: Using 'entropy' measures information gain, potentially creating more informative splits that can distinguish between classes more effectively.
- class_weight: With 'balanced', the algorithm adjusts weights inversely proportional to class frequencies, ensuring that minority classes are given appropriate emphasis during training.

```python
# Evaluate on test set
rf_best_smote = rf_bayes_search_smote.best_estimator_
y_pred_rf_smote = rf_best_smote.predict(X_test)
accuracy = metrics.accuracy_score(y_test, y_pred_rf_smote)
print(f"Test set Accuracy (SMOTE): {accuracy:.4f}")
plot_metrics(y_test, y_pred_rf_smote)
plot_roc_auc(y_test, rf_best_smote.predict_proba(X_test), classes=['Normal', 'Prediabetes', 'Diabetes'], n_classes=3)
```

- SMOTE improves the model’s ability to predict the "Prediabetes" class, increasing its recall (from 0 to 0.071) and precision (from 0 to 0.113). While still low, this is an improvement over the previous model.
- Accuracy decreases slightly to 83.44%, likely due to noise introduced by oversampling. However, the F1-scores for the "Normal" and "Diabetes" classes remain stable (0.911 and 0.711, respectively).
- The AUC for "Normal" decreases marginally (from 0.87 to 0.86), while the "Prediabetes" AUC drops to 0.64 due to oversampling-induced noise. The AUC for "Diabetes" remains constant at 0.93.

- Impact of SMOTE: SMOTE partially alleviates the class imbalance issue by improving the "Prediabetes" class recall and precision. However, the gains are modest and introduce slight performance degradation for the majority classes ("Normal" and "Diabetes").
- Trade-offs: While SMOTE enhances the minority class metrics, it reduces overall accuracy and AUC for "Prediabetes." This reflects a common trade-off in imbalanced datasets—improving minority class performance often affects majority class performance.

### 4.5 Model 4: Neural Network

Neural networks (NN) can model complex, nonlinear relationships. They are innovative in many classification tasks, particularly those involving large and multidimensional datasets.

Architecture:
- Input Layer: Number of neurons equals the number of features.
- Hidden Layers: Experimenting with different configurations.
- Activation Functions: ReLU for hidden layers, Softmax for output layer.
- Output Layer: Three neurons corresponding to the three classes.


Hyperparameters:
- Learning Rate: Experimenting with 0.001 and 0.0005.
- Batch Size: 32
- Epochs: Up to 100 with Early Stopping
- Optimizer: Adam
- Loss Function: Categorical Crossentropy

#### 4.5.1 Experimenting with Architecture and Activation Functions

```python
# Convert target variable to categorical
y_train_cat = to_categorical(y_train, num_classes=3)
y_val_cat = to_categorical(y_val, num_classes=3)
y_test_cat = to_categorical(y_test, num_classes=3)

# Define function to create NN model
def create_nn_model(input_dim, neurons_layer1, neurons_layer2, activation='relu', learning_rate=0.001):
    model = Sequential()
    model.add(Dense(neurons_layer1, input_dim=input_dim, activation=activation))
    model.add(Dropout(0.3))
    model.add(Dense(neurons_layer2, activation=activation))
    model.add(Dropout(0.3))
    model.add(Dense(3, activation='softmax'))
    optimizer = Adam(learning_rate=learning_rate)
    model.compile(loss='categorical_crossentropy', optimizer=optimizer, metrics=['accuracy'])
    return model

# Experiment with different configurations
configurations = [
    {'neurons_layer1': 64, 'neurons_layer2': 32},
    {'neurons_layer1': 128, 'neurons_layer2': 64},
    {'neurons_layer1': 256, 'neurons_layer2': 128},
]

results = []

for config in configurations:
    print(f"Training with configuration: {config}")
    nn_model = create_nn_model(
        input_dim=X_train.shape[1],
        neurons_layer1=config['neurons_layer1'],
        neurons_layer2=config['neurons_layer2'],
        activation='relu',
        learning_rate=0.0005
    )
    early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True, verbose=0)
    history = nn_model.fit(
        X_train,
        y_train_cat,
        epochs=100,
        batch_size=32,
        validation_data=(X_val, y_val_cat),
        callbacks=[early_stop],
        verbose=0
    )
    test_loss, test_accuracy = nn_model.evaluate(X_test, y_test_cat, verbose=0)
    print(f"Test Accuracy: {test_accuracy:.4f}")
    results.append({'config': config, 'accuracy': test_accuracy})
```

Apparently ss you increase the number of neurons in both hidden layers, there's a slight improvement in test accuracy. However, the improvement from the first to the third configuration is about 0.36%, which is relatively small considering the increase in model complexity.

Select the best one:

```python
best_nn_config = max(results, key=lambda x: x['accuracy'])
print(f"Best Neural Network Configuration: {best_nn_config['config']}")
print(f"Best Test Accuracy: {best_nn_config['accuracy']:.4f}")
```

```python
plot_metrics(y_test, np.argmax(nn_model.predict(X_test), axis=1))
plot_roc_auc(y_test, nn_model.predict(X_test), classes=['Normal', 'Prediabetes', 'Diabetes'], n_classes=3)
```

Similar results to previous models. The class 'Preidabetes' is still not being predicted very well. We will try to improve this with SMOTE.

```python
# Apply SMOTE
X_resampled_nn, y_resampled_nn = smote.fit_resample(X_train, y_train)
y_resampled_nn_cat = to_categorical(y_resampled_nn, num_classes=3)

# Retrain with best configuration
nn_model_smote = create_nn_model(
    input_dim=X_resampled_nn.shape[1],
    neurons_layer1=best_nn_config['config']['neurons_layer1'],
    neurons_layer2=best_nn_config['config']['neurons_layer2'],
    activation='relu',
    learning_rate=0.0005
)

nn_model_smote.fit(
    X_resampled_nn,
    y_resampled_nn_cat,
    epochs=100,
    batch_size=32,
    validation_data=(X_val, y_val_cat),
    callbacks=[early_stop],
    verbose=0
)

# Evaluate on test set
test_loss_smote, test_accuracy_smote = nn_model_smote.evaluate(X_test, y_test_cat, verbose=0)
print(f"Neural Network with SMOTE Test Accuracy: {test_accuracy_smote:.4f}")
plot_metrics(y_test, np.argmax(nn_model_smote.predict(X_test), axis=1))
plot_roc_auc(y_test, nn_model_smote.predict(X_test), classes=['Normal', 'Prediabetes', 'Diabetes'], n_classes=3)
```

In this case it is clear that SMOTE is not improving the neural network's performance. The overall accuracy dropped to 68%, the lowest across the models. The prediction for 'Prediabetes' improved and it is slightly highest than what previous models were able to achive. Hwever the F1 score for Normal dropped significantly. This might be given by the fact that meural networks can be sensitive to overfitting with oversampled data.
